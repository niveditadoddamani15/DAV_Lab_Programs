```python
grade_info = pd.DataFrame({
    'grade': ['A','B','c','D','F'],
    'grade_description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})

print("\nGrade Information DataFrame:")
print(grade_info)
```

```python
merged = pd.merge(
df,
grade_info,
on = 'grade',
how = 'left'
)
print("\n Merged DataFrame:")
print(merged.head())
```

```python
grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ],
    'perfomance_level': [
        'Very High',
        'High',
        'Medium',
        'Low',
        'Very Low'
    ]
}, index= ['A','B','C','D','F'])
print(grade_details.head())
```

```python
grade_counts =df['grade'].value_counts().to_frame('student_count')
print(grade_counts.head())
joined = grade_counts.join(grade_details)
print(joined)
```

```python
df1 =df.iloc[:100]
df2 =df.iloc[100:200]
combined = pd.concat([df1,df2])
print("\nConcatenated DataFrame:")
print(combined.head())

print(combined.shape)
```

```python
average_score = df.groupby('grade')['total_score'].mean()
print("\n Average Total score by grade:")
print(average_score)
```

```python
pivot = pd.pivot_table(
df,
values = 'total_score',
index = 'grade',
aggfunc = 'mean')

print("\nPivot Table  = Averge score by Grade:")
print(pivot)
```
